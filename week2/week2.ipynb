## Q1. Use the NLTK Brown Corpus to perform sentence tokenization, word tokenization, and stop-word removal.

```python
#(a) Load the Brown Corpus and select at least 20 sentences.
import nltk
from nltk.corpus import brown
nltk.download('brown')

sentences = brown.sents()[:20]  
```

```python
for i, sent in enumerate(sentences):
    print(f"Sentence {i+1}: {' '.join(sent)}")
```

```python
#(b) Perform sentence and word-level tokenization.
nltk.download('punkt')
from nltk.tokenize import sent_tokenize, word_tokenize
text = ' '.join(' '.join(sent) for sent in sentences)
sentences_tokenized = sent_tokenize(text)
word_tokens = word_tokenize(text)
```

```python
for s in sentences_tokenized:
    print(f"Sentence Tokenized: {s}")
for w in word_tokens:
    print(f"Word Tokenized: {w}")
```

```python
# (c) Convert tokens to lowercase. 
lower_sentences = [word.lower() for word in word_tokens]
lower_sentences
```

```python
# (d) Remove punctuation and English stop words.
import string
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
punctuation = set(string.punctuation)

cleaned_sentences = [
    [word for word in sent if word not in stop_words and word not in punctuation] 
    for sent in lower_sentences
    ]
cleaned_sentences
```

## Q2. Apply stemming and lemmatization to words extracted from the Brown Corpus and compare the results.

```python
import pandas as pd
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download('wordnet')
# (a) Extract at least 30 alphabetic words from the Brown Corpus
words = [w.lower() for w in brown.words() if w.isalpha()]
seen = []
for w in words:
    if w not in seen:
        seen.append(w)
    if len(seen) >= 30:
        break
```

```python
# (d) Include specific target words
target_words = ['playing', 'studies', 'running', 'better', 'cars']
sample_words = seen + [w for w in target_words if w not in seen]
```

```python
# (b) Apply Porter Stemmer
stemmer = PorterStemmer()
stems = [stemmer.stem(word) for word in sample_words]
```

```python
# (c) Apply WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(word) for word in sample_words]
```

```python
# (e) Create comparison DataFrame
df = pd.DataFrame({
    'Original Word': sample_words,
    'Stemmed': stems,
    'Lemmatized': lemmas
})
print(df)
```

## Q3. Implement basic spelling correction using the NLTK Words Corpus.

```python
from nltk.corpus import words
from nltk.metrics.distance import edit_distance
nltk.download('words')
# (a) Load the NLTK words corpus
correct_words = set(words.words())
correct_words
```

```python
# (b) Intentionally misspelled words
misspelled = ['recieve', 'definately', 'seperate', 'occured', 'untill',
              'wich', 'begining', 'accomodate', 'goverment', 'neccessary']
```

```python
# (c) Approximate string matching using edit distance
def correct_word(word, vocab, max_candidates=50):
    candidates = [w for w in vocab if w[0] == word[0]] or list(vocab)
    best = min(candidates, key=lambda w: edit_distance(word, w))
    return best

suggestions = {word: correct_word(word, correct_words) for word in misspelled}
```

```python
# (d) Display misspelled words and suggestions
for wrong, suggestion in suggestions.items():
    print(f"Misspelled: {wrong} -> Suggested Correction: {suggestion}")
```

```python
# (e) Correct at least five example sentences
sentences = [
    "I will recieve the package tommorow.",
    "This is definately the wrong answer.",
    "Please seperate the items into two piles.",
    "The event occured last night.",
    "Wait untill the meeting starts."
]

def correct_sentence(sentence, corrections):
    words_in_sentence = sentence.replace('.', '').split()
    corrected = [corrections.get(w.lower(), w) for w in words_in_sentence]
    return ' '.join(corrected)

for s in sentences:
    print("Original: ", s)
    print("Corrected:", correct_sentence(s, suggestions))
    print()
```

## Q4. Use the NLTK Brown Corpus to generate and analyze unigram, bigram, and trigram sequences. 

```python
from nltk import ngrams
from nltk import FreqDist
# (a) Load a suitable subset of the Brown Corpus
subset = brown.words(categories='news')
```

```python
# (b) Retain lowercase alphabetic tokens
tokens = [w.lower() for w in subset if w.isalpha()]
```

```python
# (c) Generate unigrams, bigrams, and trigrams
unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))
```

```python
# (d) Calculate N-gram frequencies
unigram_freq = FreqDist(unigrams)
bigram_freq = FreqDist(bigrams)
trigram_freq = FreqDist(trigrams)
```

```python
#(e) Display the ten most frequent unigrams, bigrams, and trigrams.
print("Most Frequent Unigrams:")
for unigram, freq in unigram_freq.most_common(10):
    print(f"{unigram[0]}: {freq}")
print("\nMost Frequent Bigrams:")
for bigram, freq in bigram_freq.most_common(10):
    print(f"{bigram[0]} {bigram[1]}: {freq}")
print("\nMost Frequent Trigrams:")
for trigram, freq in trigram_freq.most_common(10):
    print(f"{trigram[0]} {trigram[1]} {trigram[2]}: {freq}")
```

### (f) Explain how increasing N captures more contextual information
As N increases, each n-gram captures more surrounding context: unigrams only reflect word frequency in isolation (e.g., the is common everywhere), bigrams capture local word-pair relationships (e.g., of the, in the) that hint at syntax, and trigrams capture short phrase-level patterns (e.g., one of the) that reveal more about word order and collocation. However, higher-order n-grams also become sparser — each specific trigram occurs far less often than each unigram — so they need more data to estimate reliably and generalize worse to unseen text.
