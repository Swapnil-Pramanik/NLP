## Q1. Use NLTK WordNet to explore synsets and meanings of selected English words.

```python
# (a) Download and load the WordNet corpus using NLTK.
import nltk
from nltk.corpus import wordnet as wn
nltk.download('wordnet')
```

```python
# (b) Select at least 10 English words such as car, bank, bat, dog, light, plant, book, teacher, student, andcomputer.
words = ['car', 'bank', 'bat', 'dog', 'light', 'plant', 'book', 'teacher', 'student', 'computer']
```

```python
# (c) Display all or selected synsets for each word.
for word in words:
    synsets = wn.synsets(word)
    print(f"Synsets for '{word}':")
    for synset in synsets:
        print(f"  {synset.name()}")
    print()
```

```python
# (d) Display the synset name, definition, and example sentence where available.
for word in words:
    synsets = wn.synsets(word)
    print(f"Synsets for '{word}':")
    for synset in synsets:
        print(f"  {synset.name()} - {synset.definition()}")
        examples = synset.examples()
        if examples:
            print(f"    Example: {examples[0]}")
    print()
```

## Q2. Use WordNet to identify hypernyms and analyze hierarchical semantic relationships.

```python
# (a) Select at least 10 nouns from WordNet
nouns = ['car', 'bank', 'bat', 'dog', 'light', 'plant', 'book', 'teacher', 'student', 'computer']
```

```python
# (b) Identify a suitable synset for each word.
for word in nouns:
    synsets = wn.synsets(word, pos=wn.NOUN)
    if synsets:
        print(f"Selected synset for '{word}': {synsets[0].name()} - {synsets[0].definition()}")
    else:
        print(f"No synset found for '{word}'")
```

```python
# (c) Extract its direct hypernyms.
for word in nouns:
    synsets = wn.synsets(word, pos=wn.NOUN)
    if synsets:
        synset = synsets[0]
        hypernyms = synset.hypernyms()
        if hypernyms:
            print(f"Direct hypernyms for '{word}':")
            for hypernym in hypernyms:
                print(f"  {hypernym.name()} - {hypernym.definition()}")
        else:
            print(f"No direct hypernyms found for '{word}'")
    else:
        print(f"No synset found for '{word}'")
```

```python
#(d) For at least five words, trace the hypernym hierarchy for two or more levels
for word in nouns[:5]:
    synsets = wn.synsets(word, pos=wn.NOUN)
    if synsets:
        synset = synsets[0]
        print(f"Hypernym hierarchy for '{word}':")
        level = 0
        while synset.hypernyms() and level < 2:
            hypernym = synset.hypernyms()[0]
            print(f"  Level {level + 1}: {hypernym.name()} - {hypernym.definition()}")
            synset = hypernym
            level += 1
    else:
        print(f"No synset found for '{word}'")
```

```python
# (e) Display the word, synset, and hypernym chain in a Pandas DataFrame.
data = []
for word in nouns[:5]:
    synsets = wn.synsets(word, pos=wn.NOUN)
    if synsets:
        synset = synsets[0]
        level = 0
        while synset.hypernyms() and level < 2:
            hypernym = synset.hypernyms()[0]
            data.append({
                "Word": word,
                "Synset": synset.name(),
                "Hypernym": hypernym.name()
            })
            synset = hypernym
            level += 1

df = pd.DataFrame(data)
print(df)
```

## Q3. Calculate WordNet semantic similarity between different word pairs.

```python
# (a) Select at least eight word pairs, including semantically similar and dissimilar words
word_pairs = [
    ("car", "vehicle"),
    ("dog", "animal"),
    ("light", "lamp"),
    ("book", "novel"),
    ("teacher", "professor"),
    ("student", "pupil"),
    ("computer", "laptop"),
    ("bank", "financial_institution")
]
```

```python
# (b) Obtain appropriate WordNet synsets for each word.
synset_pairs = []
for word1, word2 in word_pairs:
    synsets1 = wn.synsets(word1, pos=wn.NOUN)
    synsets2 = wn.synsets(word2, pos=wn.NOUN)
    if synsets1 and synsets2:
        synset_pairs.append((synsets1[0], synsets2[0]))
```

```python
# (c) Calculate WordNet path similarity for each pair.
path_similarities = []
for synset1, synset2 in synset_pairs:
    similarity = synset1.path_similarity(synset2)
    path_similarities.append(similarity)
```

```python
# (d) Compare pairs such as car–automobile, dog–cat, dog–animal, and dog–computer.
for i, (word1, word2) in enumerate(word_pairs):
    print(f"Similarity between '{word1}' and '{word2}': {path_similarities[i]}")
```

```python
# (e) Store the results in a Pandas DataFrame and rank the pairs according to similarity score.
df = pd.DataFrame({
    "Word Pair": [f"{word1}–{word2}" for word1, word2 in word_pairs],
    "Similarity": path_similarities
})
df = df.sort_values("Similarity", ascending=False)
print(df)
```

## Q4. Analyze homonyms and contextual words to demonstrate the importance of context in determining meaning.

```python
# (a) Select at least five homonyms such as bank, bat, bark, light, and match
homonyms = ["bank", "bat", "bark", "light", "match"]
```

```python
# (b) Create at least two sentences for each word, where the word has different meanings.
sentences = {
    "bank": [
        "I sat on the bank of the river.",
        "I went to the bank to deposit my money."
    ],
    "bat": [
        "The bat flew around the cave.",
        "He swung the bat and hit the ball."
    ],
    "bark": [
        "The dog barked loudly.",
        "The tree has rough bark."
    ],
    "light": [
        "The light in the room was dim.",
        "He turned on the light switch."
    ],
    "match": [
        "We had a close match.",
        "He struck a match to light the candle."
    ]
}
```

```python
# (c) Use WordNet to display different synsets corresponding to the homonyms.
for word in homonyms:
    synsets = wn.synsets(word)
    print(f"Synsets for '{word}':")
    for synset in synsets:
        print(f"  {synset.name()} - {synset.definition()}")
    print()
```

```python
# (d) Compare the contextual sentences and manually identify the intended meaning.
for word in homonyms:
    print(f"Contextual sentences for '{word}':")
    for sentence in sentences[word]:
        print(f"  {sentence}")
    print()
```

```python
# (f) Create a final table containing the word, sentence, intended meaning, and relevant contextual clue
table_data = []
for word in homonyms:
    for i, sentence in enumerate(sentences[word]):
        # Manually identify the intended meaning and contextual clue
        intended_meaning = input(f"Enter the intended meaning for '{sentence}': ")
        contextual_clue = input(f"Enter the relevant contextual clue for '{sentence}': ")
        table_data.append({
            "Word": word,
            "Sentence": sentence,
            "Intended Meaning": intended_meaning,
            "Contextual Clue": contextual_clue
        })

df = pd.DataFrame(table_data)
print(df)
```
